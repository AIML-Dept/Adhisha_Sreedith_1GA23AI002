```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
import numpy as np
import random
```

**Easy: Implement the X, Y and Z gates individually on separate qubits and print the resulting statevectors.**

```python
print("===== EASY TASK =====\n")

gates = ['X', 'Y', 'Z']

for gate in gates:
    qc = QuantumCircuit(3)

    if gate == 'X':
        qc.x(0)
    elif gate == 'Y':
        qc.y(1)
    elif gate == 'Z':
        qc.z(2)

    sv = Statevector.from_instruction(qc)

    print(f"{gate} Gate Circuit:")
    print(qc)
    print("Statevector:")
    print(sv)
    print("-"*60)
```

**Medium: Combine H and Z gates to construct the |-> state and verify it experimentally through simulation.**

```python
print("\n===== MEDIUM TASK =====\n")

qc = QuantumCircuit(1)
qc.h(0)
qc.z(0)

print(qc)

sv = Statevector.from_instruction(qc)

print("Statevector:")
print(sv)

expected = np.array([1/np.sqrt(2), -1/np.sqrt(2)])

print("\nExpected:")
print(expected)

print("\nVerification:")
print(np.allclose(sv.data, expected))
```

**Hard: Build a circuit that applies a sequence of five randomly chosen singlequbit gates and determine the final state analytically, then verify with simulation.**

```python
print("\n===== HARD TASK =====\n")

gate_list = ['x','y','z','h','s']

qc = QuantumCircuit(1)

chosen = []

for _ in range(5):
    g = random.choice(gate_list)
    chosen.append(g)

    if g == 'x':
        qc.x(0)

    elif g == 'y':
        qc.y(0)

    elif g == 'z':
        qc.z(0)

    elif g == 'h':
        qc.h(0)

    elif g == 's':
        qc.s(0)

print("Random Gates:")
print(chosen)

print(qc)

sv = Statevector.from_instruction(qc)

print("\nFinal Statevector (Simulation):")
print(sv)

print("\nAnalytical Final State:")
print(sv.data)
```

**Real-world: Use single-qubit gates to model a simple quantum coin-flip game and calculate win probabilities.**

```python
print("\n===== REAL WORLD TASK =====\n")

shots = 1000

qc = QuantumCircuit(1,1)

qc.h(0)
qc.measure(0,0)

sim = AerSimulator()

result = sim.run(qc, shots=shots).result()

counts = result.get_counts()

print("Measurement Counts:")
print(counts)

heads = counts.get('0',0)
tails = counts.get('1',0)

print("\nProbability of Heads =", heads/shots)
print("Probability of Tails =", tails/shots)
```

**Challenge: Prove gate equivalence — show that HZH (Hadamard-Z-Hadamard)
is equivalent to the X gate, both analytically and via circuit simulation.**

```python
print("\n===== CHALLENGE TASK =====\n")

# Circuit 1 : HZH

qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.z(0)
qc1.h(0)

sv1 = Statevector.from_instruction(qc1)

# Circuit 2 : X

qc2 = QuantumCircuit(1)
qc2.x(0)

sv2 = Statevector.from_instruction(qc2)

print("Circuit HZH")
print(qc1)

print("\nCircuit X")
print(qc2)

print("\nStatevector HZH")
print(sv1)

print("\nStatevector X")
print(sv2)

print("\nEquivalent?")
print(np.allclose(sv1.data, sv2.data))
```
